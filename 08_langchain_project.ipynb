```python
from google.colab import userdata
import os
import dotenv

from langchain import PromptTemplate
from langchain import hub
from langchain.docstore.document import Document
from langchain.document_loaders import WebBaseLoader
from langchain.schema import StrOutputParser
from langchain.schema.prompt_template import format_document
from langchain.schema.runnable import RunnablePassthrough
from langchain.vectorstores import Chroma
```

```python
# @title Create .env file
file_name = "./gemini.env"
with open(file_name, "w") as file:
    # Before get api_key, create a secret variable named GOOGLE_API_KEY and initialize with Open API key.
    api_key = userdata.get('GOOGLE_API_KEY')
    file.write(f"GOOGLE_API_KEY={api_key}")
```

```python
# @title Load OpenAI API key
# Write your code here
%load_ext dotenv
%dotenv ./gemini.env

GOOGLE_API_KEY=os.environ['GOOGLE_API_KEY']
```

```python
# @title Load Document
from langchain.document_loaders import TextLoader

loader = TextLoader("/content/pg1342.txt")
data = loader.load()
```

```python
# @title Split document
from langchain.text_splitter import RecursiveCharacterTextSplitter, CharacterTextSplitter

chunk_size =100
chunk_overlap = 25

r_splitter = RecursiveCharacterTextSplitter(
    chunk_size=chunk_size,
    chunk_overlap=chunk_overlap
)
```

```python
textSplit = r_splitter.split_documents(data)

print(textSplit)
```

```python
# @title Load Embedding
from langchain_google_genai import GoogleGenerativeAIEmbeddings

gemini_embeddings = GoogleGenerativeAIEmbeddings(google_api_key=GOOGLE_API_KEY, model="models/embedding-001")
```

```python
# @title Save Embedding
vectorstore = Chroma.from_documents(
                     documents=textSplit,                 # Data
                     embedding=gemini_embeddings,    # Embedding model
                     persist_directory="./chroma_db" # Directory to save data
                     )
```

```python
# @title Load Embedding
vectorstore_disk = Chroma(
                        persist_directory="./chroma_db",       # Directory of db
                        embedding_function=gemini_embeddings   # Embedding model
                   )

retriever = vectorstore_disk.as_retriever(search_kwargs={"k": 500})
print(len(retriever.get_relevant_documents("MMLU")))
```

```python
# @title Load Model
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(google_api_key=GOOGLE_API_KEY, model="gemini-pro",
                 temperature=0.75, top_p=0.85)
```

```python
# @title Create Prompt Template
llm_prompt_template = """You are an assistant for question-answering tasks.
Use the following context to answer the question.
If you don't know the answer, just say that you don't know.
Use 50 sentences maximum and keep the answer concise.\n
Question: {question} \nContext: {context} \nAnswer:"""

llm_prompt = PromptTemplate.from_template(llm_prompt_template)

print(llm_prompt)
```

```python
# @title Combine data from documents to readable string format.
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

qa = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | llm_prompt
    | llm
    | StrOutputParser()
)
```

## Query Space

```python
query = "What are the full names of the two main characters in Pride and Prejudice ?"
answer = qa.invoke(query)

question_id = "00_langchain_query_answer"
submit(student_id, name, assignment_id, str(answer), question_id, drive_link)
```
