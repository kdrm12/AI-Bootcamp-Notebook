```python
# deskripsi tentang kategori

category_description = {
    'Perfect Fit': ''' A very high level of alignment with the job description. Candidates in this range are likely to be the top performers and should be prioritized for interviews or further evaluations. ''',
    'Strong Match' : ''' A strong match with the job requirements, with only minor gaps. These candidates are highly suitable but may require some additional preparation or upskilling in specific areas. ''',
    'Partial Match': ''' A moderate match with the job description. While these candidates show potential, they may need significant improvements or further qualifications to meet the role's expectations. ''',
    'Poor Match': ''' A poor match with the job requirements. These candidates lack most of the necessary skills or experience, making them unsuitable for the position in its current form. '''
}
```

###Gradio Interface App

```python
import gradio as gr
from gradio_pdf import PDF
from pypdf import PdfReader
from g4f.client import Client

# import client LLM
client = Client()

# memuat deskripsi dari pekerjaan (job)
def load_default_job_description(job_type):
    if job_type in job_descriptions:
        return job_descriptions[job_type]
    else:
        return None

# untuk parsing resume
def resume_parser(resume_link):
    reader = PdfReader(resume_link)
    resume_text = ""
    for page in reader.pages:
        resume_text += page.extract_text() + "\n"
    return resume_text

# untuk menghitung kesesuaian score (score_suitability)
def scoring_suitability(evaluation):

    prompt = f"""
    You are a recruitment assistant. Give evaluate suitabilty score from the following evaluation.

    Evaluation:
    {evaluation}

    !!! IMPORTANT !!!
    Just respond with a numeric number in range 0 to 100

    @@ EXAMPLES @@
    EXAMPLE: 52
    EXAMPLE: 100
    EXAMPLE: 0

    ### REMEMBER ###
    Just respond with a numeric number in range 0 to 100

    @@ EXAMPLES @@
    EXAMPLE: 72
    EXAMPLE: 63
    EXAMPLE: 20
    """

    response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
    suitability_score = response.choices[0].message.content

    return suitability_score

#  untuk mengkategorisasi nilai kesesuaian (score suitability)
def score_categorization(score):
    score = int(score)
    if score >= 90:
        return "Perfect Fit"
    elif 75 <= score < 90:
        return "Strong Match"
    elif 50 <= score < 75:
        return "Partial Match"
    else:
        return "Poor Match"

# memuat deskripsi kategori
def load_category_description(category):
    description = category_description[category]
    return description

# untuk memuat proses evaluasi
def evaluation_process(resume, job_description):

    if resume is None or job_description is None:
        return "Please provide both resume and job description."

    # proses menggunakan LLM dengan meminta kelebihan dan kekurangan dari pelamar
    prompt = f"""
    You are a recruitment assistant. Evaluate the following resume for its suitability for the given job description.

    Job Description:
    {job_description}

    Resume:
    {resume}

    !!! IMPORTANT !!!
    RESPOND IN THIS FORMAT:
    Strengths: [list of strengths]
    Weaknesses: [list of weaknesses]

    MUST FOLLOWING THIS RULES:
    2. Strengths: [key strengths based on the resume]
    3. Weaknesses: [key weaknesses or missing criteria]

    ### REMEMBER ###
    RESPOND IN THIS FORMAT:
    Strengths: [list of strengths]
    Weaknesses: [list of weaknesses]
    """

    response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
    evaluation = response.choices[0].message.content

    return evaluation

# program utama
def main_program(resume_link, job_description):
    resume_text = resume_parser(resume_link)
    evaluation = evaluation_process(resume_text, job_description)
    suitabilty_score = scoring_suitability(evaluation)
    category = score_categorization(suitabilty_score)
    description = load_category_description(category)
    return suitabilty_score, category, description, evaluation

with gr.Blocks() as iface:
    gr.Markdown("# 📝 Resume Screener")
    gr.Markdown("Upload your candidate employee resume and provide job description to evaluate its suitability.")
    with gr.Row():
        with gr.Column(scale=1):
            resume = PDF(label="Resume (Pdf Only)")
        with gr.Column(scale=1):
            job_type = gr.Radio(["Accountant", "UI Designer", "AI Engineer", "Data Scientist", "Other"], label="Job Type")
            job_description = gr.Textbox(label="Job Description")
            job_type.change(load_default_job_description, inputs=job_type, outputs=job_description)
            submit_btn = gr.Button("Submit")

    with gr.Row():
      with gr.Column():
        gr.Markdown("## Results")
        with gr.Row():
            score_output = gr.Label(label="Suitability Score")
            category_output = gr.Label(label="Score Category")
        gr.Markdown("### Category Description")
        category_description_output = gr.Markdown(container = True, min_height = 50, max_height=150)
        gr.Markdown("### Evaluation")
        evaluation_output = gr.Markdown(container = True, show_copy_button = True, min_height = 450, max_height=450)


    submit_btn.click(
        fn=main_program,
        inputs=[resume, job_description],
        outputs=[score_output, category_output, category_description_output, evaluation_output]
    )

iface.launch()
```
